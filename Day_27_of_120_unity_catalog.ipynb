### Creating a table

```python
%sql
create table training_db.employees
as
select *
from read_files('/Volumes/training/training_db/training_volume/employees.csv')
```

```python
%sql
select *
from training.training_db.employees;
```

### Creating a table using spark using SAVE

```python
df = spark.read.csv('/Volumes/training/training_db/training_volume/employees.csv', header=True, inferSchema=True)

df.write.format('delta')\
    .mode("append")\
        .save('/Volumes/training/training_db/training_volume/employees_delta_table')
```

```python
df = spark.read.csv('/Volumes/training/training_db/training_volume/employees.csv', header=True, inferSchema=True)

df.write.format('delta')\
    .mode("append")\
        .saveAsTable('training.training_db.employees_delta_table')
```

```python
%sql
select *
from delta.`/Volumes/training/training_db/training_volume/employees_delta_table`;
```
