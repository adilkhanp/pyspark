## Add  a column to the dataframe
-   Adds a column if it is not present
-   Updates the value of a column if present

```python
from pyspark.sql import functions as F
from pyspark.sql.functions import col,lit
emp = emp.withColumn("annual_salary", lit(10000) )
```

```python
emp = emp.withColumns({"annual_salary": lit(10000) ,
                      "avg_salary" : lit (100000),
                      "commission" : lit("Yes")})
```

## Changing the datatype

```python
emp = emp.withColumn("salary" , col("salary").cast("integer"))
```

## Rename the column

```python
emp = emp.withColumnRenamed("annual_salary","yearly_compensation")
```

```python
emp = emp.withColumnsRenamed({"avg_salary" : "avg_compensation",
                              "PHONE_NUMBER" : "MOBILE"})
```

## DROP THE COLUMNS

```python
emp = emp.drop("yearly_compensation","avg_compensation")
```

## selecting the columns

```python
emp.select("COMMISSION_PCT","FIRST_NAME","LAST_NAME").display()
```

```python
emp.select("FIRST_NAME","LAST_NAME", (col("SALARY") * 12).alias("Annual_salary") ).display()
```

## Filtering the data

```python
emp.filter(col("DEPARTMENT_ID") == 30).display()
```

## Multiple conditions (Logical operators)

```python
emp.filter((col("DEPARTMENT_ID") == 80) & (col("salary") > 5000)).display()
```

```python
emp.filter((col("DEPARTMENT_ID") == 10) | (col("DEPARTMENT_ID") == 30)).display()
```

```python
emp.filter(( (col("DEPARTMENT_ID") == 10 ) | (col("DEPARTMENT_ID") == 30)) & (col("salary") > 3000)).display()
```

```python
emp.filter(( (col("DEPARTMENT_ID") == 10 ) | (col("DEPARTMENT_ID") == 30)) & (col("salary") > 3000)).select("FIRST_NAME","LAST_NAME",col("salary").alias("COMPENSATION"),"DEPARTMENT_ID").display()
```
