```python
df = spark.read.format('csv')\
    .option('header', 'true')\
    .option('inferSchema', 'true')\
    .load('/Volumes/training/training_db/training_volume/employees.csv')
```

```python
from pyspark.sql.functions import *

df.select( upper(col('FIRST_NAME')).alias("FIRST_NAME"),
          lower(col('LAST_NAME')).alias("LAST_NAME"),
          initcap(col('EMAIL')).alias("EMAIL")
          ).display()
```

```python
df.select(substring(col("FIRST_NAME"),1,2)).display()
```

```python
df.filter(upper(substring(col("FIRST_NAME"),1,1))=="A")\
    .select("FIRST_NAME","LAST_NAME")\
    .display()
```

```python
df.select(instr(col('PHONE_NUMBER'),'.') ,'PHONE_NUMBER'
          ).display()
```

```python
df.select(instr(col('PHONE_NUMBER'),'.') ,'PHONE_NUMBER'
          ).display()
```

```python
df.select(length('FIRST_NAME')).display()
```

```python
df.filter(length(col('FIRST_NAME')) == 6).select("FIRST_NAME","EMPLOYEE_ID").display()
```

```python
dd = [(" PYSPARK ") , (" PYTHON ")]

df1 = spark.createDataFrame(dd,['name'])

df1.select( trim(col('name')).alias("name"), rtrim(col('name')).alias("rname") , ltrim(col('name')).alias("lname")).display()
```

```python
df1.select(lpad(col('name'),10, '@').alias('lpad_name'),
           rpad(col('name'),10, '@').alias('rpad_name')).display()
```

```python
df2 = df1.select( concat(col('name'),lit(' is a good language'),lit(" yes")).alias("data"))
```

```python
df2.select(replace(col('data'),lit('good language'),lit('good programming language'))).display()
```

```python
df2.select(replace(col('data'),lit('good language'))).display()
```

```python
df.select ( upper(col('FIRST_NAME')).alias('FIRST_NAME'),
           lower(col('LAST_NAME')).alias("LAST_NAME"),
           initcap(col('EMAIL')).alias("EMAIL"),
           ltrim(col('JOB_ID')).alias("JOB_ID"))\
               .write.mode("append")\
                   .saveAsTable("training.training_db.employee_clean")
```
