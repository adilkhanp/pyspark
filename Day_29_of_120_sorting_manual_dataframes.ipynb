### **Creating DataFrame from manual data**

```python
data = [ (1 , "Adilkhan" , 50000),
         (2, "karthik", 60000),
         (3,"khayyum" , 70000),
         (4,"Nadeem" , 80000)
]

schema = [ "id" , "name" , "salary"]

df = spark.createDataFrame(data,schema)

df.printSchema()
```

```python
df.display()
```

```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType

data = [ (1 , "Adilkhan" , 50000),
         (2, "karthik", 60000),
         (3,"khayyum" , 70000),
         (4,"Nadeem" , 80000)
]

schema = StructType([StructField("id", IntegerType(), False),
                     StructField("name", StringType(), True),
                     StructField("salary", IntegerType(), True)])

df = spark.createDataFrame(data,schema)

df.printSchema()
```

### **Sorting**

```python
df = spark.read.format('csv')\
  .option('header', 'true')\
  .option('inferSchema', 'true')\
  .load('/Volumes/training/training_db/training_volume/employees.csv')
```

### Ascending Order

```python
from pyspark.sql.functions import col

df.sort(col('SALARY')).display()
```

### Descending Order

```python
df.sort(col('SALARY'),ascending = False).display()
```

### Sorting on multiple columns

```python
df.sort([col('DEPARTMENT_ID'),col('FIRST_NAME')] , ascending = [True,False]).display()
```

### Sorting and selecting only some columns

```python
df.select([col('FIRST_NAME'),col('LAST_NAME'),col('HIRE_DATE'),col('DEPARTMENT_ID')])\
    .sort(col('SALARY'))\
        .display()
```

### Sorting with Filter clause

```python
df.filter(col('DEPARTMENT_ID').isin([10,20,80,90]))\
    .sort(col('FIRST_NAME'))\
        .select([col('FIRST_NAME'),col('LAST_NAME'),col('HIRE_DATE'),col('DEPARTMENT_ID')])\
            .display()
```

## String Functions

```python
from pyspark.sql.functions import *

df.select(upper(col('FIRST_NAME')).alias("FIRST_NAME"),lower(col('LAST_NAME')).alias("LAST_NAME") \
    ,initcap(col('EMAIL')).alias("EMAIL"))\
        .display()
```
