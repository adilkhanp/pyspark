```python
df = spark.read.csv('/Volumes/training/training_db/training_volume/employees.csv', header=True)
```

```python
data = [10.233333,20.97856778,36.086547]

df1 = spark.createDataFrame(data,"double")

df1.display()
```

```python
from pyspark.sql.functions import *

df1.select(round(col('value'),2), round(col('value'),-1)).display()
```

```python
df1.select(round(pmod(col('value'),3),2)).display()
```

```python
df1.select(floor(col('value')).alias("floor"),ceil(col('value')).alias("ceil")).display()
```

```python
df1.select(sqrt(col('value')).alias("SQRT"),
           power(col('value'),3).alias("POWER")).display()
```

```python
from pyspark.sql.types import *

df2 = [ ( 10,20,30) , ( 30,10, 60 ) , ( 20 , 80 , 32 )]

schema = StructType([ StructField("c1", IntegerType(), True), StructField("c2", IntegerType(), True), StructField("c3", IntegerType(), True)])

df2 = spark.createDataFrame(df2,schema)
```

```python
df2.select ( greatest (df2.c1, df2.c2, df2.c3).alias("GREATES"),
            least(df2.c1, df2.c2, df2.c3).alias("LEAST") ).display()
```

### ** Date Functions **
- sysdate,current_date   -- current_date()
- current_timestamp , systimestamp -- current_timestamp()
- add_months
- months_between
- round
- trunc
- last_day
- next_day

```python
df2.select( current_date().alias("Today_date"),
           current_timestamp().alias("Time Now"))\
               .limit(1).display()
```

```python
df = df.withColumn("HIRE_DATE" , to_date(col("HIRE_DATE"),"dd-MMM-yy"))
df.display()
```

```python
df.select(col('HIRE_DATE') , add_months(col('HIRE_DATE'), 2).alias("2months_added"),
          add_months(col('HIRE_DATE'),-2).alias("2months_subtracted")).display()
```

```python
df.select(next_day(current_date() , 'FRI'), last_day(current_date())).limit(1).display()
```

```python
df.select(col('EMPLOYEE_ID') , col('FIRST_NAME') ,(months_between(current_date() , col('HIRE_DATE'))/12).alias("Experience_in_months")).display()
```

```python
df.select( col('HIRE_DATE'), year(col('HIRE_DATE')) , day(col('HIRE_DATE')) , month(col('HIRE_DATE'))).display()
```

```python
df.select( col('HIRE_DATE') , trunc(col('HIRE_DATE'),'MM').alias("MONTH_TRUNC"),
          trunc(col('HIRE_DATE'),'yy').alias('YEAR_TRUNC')).display()
```
