### NULL Functions
- isNa
- dropNa
- fillNa
- isNull
- isNotNull
- coalesce

```python
schema = '''EMPLOYEE_ID int,
FIRST_NAME string,
LAST_NAME string,
EMAIL string,
PHONE_NUMBER string,
HIRE_DATE string,
JOB_ID string,
SALARY int,
COMMISSION_PCT int,
MANAGER_ID int,
DEPARTMENT_ID int,
LAST_EMPLOYER string'''

df = spark.read.format('csv')\
    .option("header","true")\
        .schema(schema)\
        .load('/Volumes/training/training_db/training_volume/employees.csv')
```

```python
df = df.withColumn("HIRE_DATE", to_date(col("HIRE_DATE"),'dd-MMM-yy'))
```

```python
df.display()
```

```python
df.filter( col('COMMISSION_PCT').isNull())\
    .select("FIRST_NAME","COMMISSION_PCT")\
        .display()
```

```python
df.filter( col('MANAGER_ID').isNull())\
    .select("FIRST_NAME","COMMISSION_PCT","MANAGER_ID")\
        .display()
```

```python
df.dropna(subset=["MANAGER_ID"]
          ).display()
```

```python
df.dropna().display()
```

```python
df.filter( col('MANAGER_ID').isNotNull()).display()
```

```python
df.fillna('adilkhan').display()
```

```python
df.select ( coalesce(col("MANAGER_ID"),lit('-1'))).display()
```
