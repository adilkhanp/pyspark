```python

df = spark.read.format('csv')\
    .option('header', 'true')\
    .option('inferSchema', 'true')\
    .load('/Volumes/training/training_db/training_volume/employees.csv')
#
```

```python
from pyspark.sql.functions import *
df = df.withColumn("HIRE_DATE", to_date(col("HIRE_DATE"),'dd-MMM-yy'))
```

```python
df.groupBy(col('DEPARTMENT_ID')).count().display()
```

```python
df.groupBy(col('DEPARTMENT_ID'))\
    .sum('SALARY')\
        .alias("TOTAL_SALARY")\
        .display()
```

```python
df.groupBy(col('DEPARTMENT_ID'))\
    .agg(sum(col('SALARY')).alias("TOTAL_SALARY"),
         round(avg(col('SALARY')),2).alias("AVERAGE_SALARY"),
         min(col("SALARY")).alias('MIN_SALARY'),
         max(col("SALARY")).alias('MAX_SALARY'),
         count(col('SALARY')).alias('COUNT_SALARY')
    )\
        .display()
```

```python
df.groupBy([col('DEPARTMENT_ID'),col('JOB_ID')])\
    .agg(sum(col('SALARY')).alias("TOTAL_SALARY"),
         round(avg(col('SALARY')),2).alias("AVERAGE_SALARY"),
         min(col("SALARY")).alias('MIN_SALARY'),
         max(col("SALARY")).alias('MAX_SALARY'),
         count(col('SALARY')).alias('COUNT_SALARY')
    )\
        .display()
```

```python
df.groupBy( year(col('HIRE_DATE')).alias('Employee_Year'))\
    .agg( count('SALARY').alias('TOTAL_COUNT'))\
        .display()
```

```python
df.groupBy( upper(substring(col('FIRST_NAME'),1,1)).alias('ALPHABET'))\
    .agg( count('SALARY').alias('TOTAL_COUNT'))\
        .sort('ALPHABET')\
        .display()
```

```python
df.groupBy(col('DEPARTMENT_ID'))\
    .agg(sum('SALARY').alias('TOTAL_SALARY'),
         count('COMMISSION_PCT').alias('EARNING_COMMISSION'))\
        .sort('DEPARTMENT_ID')\
            .display()
```

```python
df.groupBy(col('DEPARTMENT_ID'))\
    .agg(sum('SALARY').alias('TOTAL_SALARY'),
         count('COMMISSION_PCT').alias('EARNING_COMMISSION'))\
             .filter(col('EARNING_COMMISSION') > 0)\
        .sort('DEPARTMENT_ID')\
            .display()
```

### *** JOINS ***


```python
emp = df
```

```python
emp.display()
```

```python
dept = spark.read.format('csv')\
    .option('header','true')\
    .option('inferSchema','true')\
    .load('/Volumes/training/training_db/training_volume/departments.csv')
```

```python
emp.join(dept, on = 'DEPARTMENT_ID', how = 'inner')\
    .display()
```

```python
emp.join(dept, emp.DEPARTMENT_ID == dept.DEPARTMENT_ID,
          how = 'inner')\
    .display()
```

```python
dept_ids = dept.select(col('DEPARTMENT_ID'))
dept_ids.display()
```

```python
emp.filter(col('DEPARTMENT_ID').isin(dept_ids)).display()
```
