```python
dd3 = [(1, 'SOMEONE', '98658,88547136'),
        (2, 'SOMETWO', '58756,89542,9658')
]

df3 = spark.createDataFrame(dd3,['id', 'name', 'phone'])
df3.display()
```

```python
from pyspark.sql.functions import *
df3.select( col('id') , col('name'), split(col('phone'),lit(','))[0]).display()
```

```python
df4 = df3.select( col('id') , col('name'), explode(split(col('phone'),lit(','))).alias('phone'))
```

```python
df4.display()
```

```python
df4.groupBy('id','name')\
    .agg(collect_list('phone').alias('phone'))\
        .display()
```

```python
df4.groupBy('id','name')\
    .agg(concat_ws(lit(','),collect_list('phone')).alias('phone'))\
        .display()
```

```python
emp = spark.read.format('csv')\
    .option('header',True)\
    .option('inferSchema',True)\
    .load('/Volumes/training/training_db/training_volume/employees.csv')
```
